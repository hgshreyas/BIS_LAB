```python
import random

# User input
n = int(input("Enter number of buses: "))
m = int(input("Enter number of passengers: "))

passengers = list(map(int, input(
    "Enter passenger arrival times (minutes): "
).split()))

start = int(input("Enter start time (minutes): "))
end = int(input("Enter end time (minutes): "))

POP_SIZE = 10
GENERATIONS = 50
MUTATION_RATE = 0.1

# Fitness function
def fitness(schedule):
    waiting = 0

    for p in passengers:
        buses = [b for b in schedule if b >= p]

        if buses:
            waiting += min(buses) - p
        else:
            waiting += end - p

    return 1 / (1 + waiting)

# Initial population
population = [
    sorted(random.sample(
        range(start, end + 1), n
    ))
    for _ in range(POP_SIZE)
]

# Genetic Algorithm
for generation in range(GENERATIONS):

    population.sort(key=fitness, reverse=True)
    new_population = [population[0]]

    while len(new_population) < POP_SIZE:
        p1, p2 = random.sample(population[:5], 2)

        # Crossover
        child = sorted(random.sample(p1 + p2, n))

        # Mutation
        if random.random() < MUTATION_RATE:
            i = random.randrange(n)
            child[i] = random.randint(start, end)
            child = sorted(child)

        new_population.append(child)

    population = new_population

# Best solution
best = max(population, key=fitness)

print("\nBest Bus Departure Schedule:", best)
print("Minimum Passenger Waiting Time:",
      round(1 / fitness(best) - 1), "minutes")
```
