```python
import random

# User input
n = int(input("Enter number of cities: "))

print("Enter distance matrix:")
dist = []

for i in range(n):
    row = list(map(float, input().split()))
    dist.append(row)

# ACO parameters
ants = 10
iterations = 50
alpha = 1
beta = 2
rho = 0.5
Q = 100

# Initialize pheromone
pheromone = [[1.0 for j in range(n)] for i in range(n)]

# Calculate route distance
def route_distance(route):
    total = 0

    for i in range(n):
        total += dist[route[i]][route[(i + 1) % n]]

    return total

# Construct route for each ant
def create_route():
    route = [0]
    unvisited = list(range(1, n))

    while unvisited:
        current = route[-1]
        probabilities = []

        for city in unvisited:
            tau = pheromone[current][city] ** alpha
            eta = (1 / dist[current][city]) ** beta
            probabilities.append(tau * eta)

        total = sum(probabilities)
        probabilities = [p / total for p in probabilities]

        next_city = random.choices(
            unvisited, weights=probabilities
        )[0]

        route.append(next_city)
        unvisited.remove(next_city)

    return route

# Best solution
best_route = None
best_distance = float('inf')

# Main ACO algorithm
for iteration in range(iterations):

    routes = []

    for ant in range(ants):
        route = create_route()
        distance = route_distance(route)

        routes.append((route, distance))

        if distance < best_distance:
            best_distance = distance
            best_route = route[:]

    # Pheromone evaporation
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # Pheromone update
    for route, distance in routes:
        for i in range(n):
            a = route[i]
            b = route[(i + 1) % n]

            pheromone[a][b] += Q / distance
            pheromone[b][a] += Q / distance

    print("Iteration", iteration + 1,
          "Shortest Distance:", round(best_distance, 2))

# Output
print("\nBest Route:", best_route + [0])
print("Shortest Distance:", round(best_distance, 2))
```
