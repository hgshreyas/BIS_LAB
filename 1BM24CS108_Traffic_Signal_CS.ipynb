```python
import random

# User input
n = int(input("Enter number of traffic signals: "))

arrival = list(map(int, input(
    "Enter vehicle arrival rates: ").split()))

queue = list(map(int, input(
    "Enter queue lengths: ").split()))

waiting = list(map(int, input(
    "Enter average waiting times: ").split()))

# Parameters
N = 10
ITERATIONS = 30
PA = 0.25

# Fitness function
def fitness(nest):
    cost = 0

    for i in range(n):
        cost += arrival[i] * waiting[i] / nest[i]
        cost += queue[i] / nest[i]

    return cost

# Initialize nests (green signal timings)
nests = [
    [random.uniform(10, 60) for _ in range(n)]
    for _ in range(N)
]

# Levy flight
def levy_flight():
    return random.gauss(0, 1) / (
        abs(random.gauss(0, 1)) ** (1 / 1.5) + 0.01
    )

# Best nest
best = min(nests, key=fitness)

# Cuckoo Search
for iteration in range(ITERATIONS):

    for i in range(N):

        new_nest = nests[i][:]

        for j in range(n):
            new_nest[j] += 2 * levy_flight()
            new_nest[j] = max(10, min(60, new_nest[j]))

        k = random.randint(0, N - 1)

        if fitness(new_nest) < fitness(nests[k]):
            nests[k] = new_nest

    # Abandon worst nests
    nests.sort(key=fitness)

    for i in range(int(PA * N)):
        nests[-(i + 1)] = [
            random.uniform(10, 60) for _ in range(n)
        ]

    current = min(nests, key=fitness)

    if fitness(current) < fitness(best):
        best = current[:]

    print("Iteration", iteration + 1,
          "Best Cost:", round(fitness(best), 2))

# Output
print("\nOptimal Signal Timings:")
for i in range(n):
    print("Signal", i + 1, ":", round(best[i], 2), "seconds")

print("Minimum Traffic Cost:", round(fitness(best), 2))
```
